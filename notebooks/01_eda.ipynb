# 01. Exploratory Data Analysis (EDA)
## AI-Driven Spatio-Temporal Tracking of Extreme Weather Anomalies
This notebook inspects the South Asia meteorological grid (5°-40°N, 60°-100°E) and analyzes distributions of temperature, precipitation, wind fields, and pressure.

```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Open preprocessed dataset
data_path = Path('../data/processed/processed_weather.nc')
ds = xr.open_dataset(data_path)
print('Dataset Dimensions:', ds.sizes)
print('Variables:', list(ds.data_vars.keys()))
```

```python
# Summary statistics for meteorological variables
for var in ['t2m', 'tp', 'u10', 'v10', 'msl', 'rh']:
    if var in ds:
        vals = ds[var].values
        print(f'{var.upper():5s} | Mean: {np.mean(vals):10.2f} | Std: {np.std(vals):8.2f} | Min: {np.min(vals):8.2f} | Max: {np.max(vals):8.2f}')
```

```python
# Plot spatial snapshot of temperature and precipitation anomalies
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
t_idx = 150 # Peak summer snapshot

ds['t2m_anomaly'].isel(time=t_idx).plot(ax=axes[0], cmap='inferno')
axes[0].set_title(f'2m Temperature Anomaly (K) - Day {t_idx}')

ds['tp_anomaly'].isel(time=t_idx).plot(ax=axes[1], cmap='Blues')
axes[1].set_title(f'Precipitation Anomaly (mm/d) - Day {t_idx}')
plt.tight_layout()
plt.show()
```
