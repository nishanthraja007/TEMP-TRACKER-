This notebook analyzes model checkpoints, loss curves, and comparative evaluation tables.

```python
import json
import pandas as pd
from pathlib import Path

# Load Ablation Study Table
ablation_path = Path('../results/ablation.csv')
if ablation_path.exists():
    df = pd.read_csv(ablation_path)
    print('=== ABLATION STUDY RESULTS ===')
    print(df.to_string(index=False))
else:
    print('Run: python -m src.evaluate')
```

```python
# Inspect Forecast Metrics
forecast_file = Path('../results/forecast_metrics.json')
if forecast_file.exists():
    with open(forecast_file, 'r') as f:
        fc = json.load(f)
    print('=== FORECAST VERIFICATION METRICS (ConvLSTM vs Baselines) ===')
    for lead in ['day_1', 'day_2', 'day_3']:
        c_rmse = fc['convlstm'][lead]['rmse']
        p_rmse = fc['persistence'][lead]['rmse']
        print(f'{lead.upper()} | ConvLSTM RMSE: {c_rmse:.4f} | Persistence RMSE: {p_rmse:.4f}')
```
