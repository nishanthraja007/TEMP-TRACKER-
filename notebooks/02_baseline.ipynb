# 02. Baseline Anomaly Detection
## Statistical Thresholds, Z-Scores & IMD Heatwave Rules
This notebook evaluates baseline statistical detection algorithms against ground truth masks.

```python
import json
from pathlib import Path
import xarray as xr
import matplotlib.pyplot as plt

metrics_file = Path('../results/baseline_metrics.json')
if metrics_file.exists():
    with open(metrics_file, 'r') as f:
        metrics = json.load(f)
    print('Baseline Detection Metrics:', json.dumps(metrics, indent=2))
else:
    print('Baseline metrics not found. Run: python -m src.anomaly.baseline')
```

```python
# Inspect Baseline Anomaly Masks
mask_path = Path('../results/baseline_masks.nc')
if mask_path.exists():
    b_ds = xr.open_dataset(mask_path)
    fig, ax = plt.subplots(figsize=(8, 6))
    b_ds['baseline_heatwave_mask'].isel(time=150).plot(ax=ax, cmap='Reds')
    ax.set_title('Baseline Heatwave Mask (Day 150)')
    plt.show()
```
